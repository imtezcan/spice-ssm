```python
import os.path
import pybeam.custom as pbc
```

```python
# Configuration

# get parent directory
dir_model = 'custom_models/get_traces'
N_sims = 1000
phi = {
    'phi[0]': 0.2,  # non-decision time
    'phi[1]': 0.5,  # relative starting point
    'phi[2]': 1.0,  # drift rate
    'phi[3]': 1.0,  # diffusion constant
    'phi[4]': 1.0,  # threshold
}
```

```python
# get likelihoods of response times
rt, traces = pbc.simulate(model_dir=dir_model, N_sims=N_sims, phi=phi, get_traces=True)
```

```python
import matplotlib.pyplot as plt

# plot some traces
fig, ax = plt.subplots()
for i in range(10):
    # get index of non-zero elements
    idx = traces[i] != 0
    ax.plot(traces[i, idx])
```
